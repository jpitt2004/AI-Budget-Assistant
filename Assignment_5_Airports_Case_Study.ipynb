```python
import pandas as pd
import seaborn as sns
```

```python
tsaClaims = pd.read_csv("tsa_claims1-2.csv", low_memory=False)
```

```python
tsaClaims.head()
```

### **1. What is the most common type of insurance claim?**
---
**Answer:** The most common type of insurance claim is Passenger Property Loss, with 60,265 claims. See Below!

```python
# Question 1
tsaClaims['Claim Type'].value_counts()
```

### **2. Which claim site within the airport are claims most commonly filed for?**
---
**Answer:** Claims are most commonly filed for Checked Baggage, with 80,553 claims. See Below!

```python
# Question 2
tsaClaims['Claim Site'].value_counts()
```

### **3. What type of claim is made most at each claim site?**
---
**Answer:** Passenger Property Loss is the most common claim type for Checked Baggage, Motor Vehicle, and Other claim sites. Property Damage is the most common claim type at Checkpoints. See Below!

```python
# Question 3
tsaClaims.groupby('Claim Site')['Claim Type'].agg(
    lambda x: x.value_counts().index[0]
)
```

### **4. What is the typical claim amount?**
---
**Answer:** The typical claim amount is about $183.69, based on the median. The median gives a better idea of a typical claim because there are some extremely high claim amounts that make the average much higher. See Below!

```python
# Question 4
pd.options.display.float_format = '{:.2f}'.format
tsaClaims['Claim Amount'].describe()
```

### **5. What is the overall claim approval rate for the entire U.S.?**
---
**Answer:** The overall claim approval rate for the United States is approximately 24.43%. See Below!

```python
# Question 5
approval_rate = (tsaClaims['Status'] == 'Approved').mean() * 100
approval_rate
```
